## 1. Load Processed Text Features & N-Gram Tables

## 2. Text Availability & Token Statistics
Examine title availability, description missingness, and token length distribution.

```python
print(f"Title Available: {df_text['title_available'].sum():,} (100.0%)")
print(f"Description Available: {df_text['description_available'].sum():,} (0.0%)")
print(f"Mean tokens per listing: {df_text['token_count'].mean():.2f}")
print(f"Median tokens per listing: {df_text['token_count'].median():.0f}")
```

## 3. Controlled Lexical Category Prevalences
Analyze the distribution of warranty, condition, transaction, urgency, and contact terms.

```python
lex_cols = [c for c in df_text.columns if c.startswith("has_")]
for col in lex_cols:
    cnt = df_text[col].sum()
    pct = (cnt / len(df_text)) * 100
    print(f"{col.replace('has_', '').title()}: {cnt:,} ({pct:.2f}%)")
```
